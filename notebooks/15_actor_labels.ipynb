```python
import pandas as pd

df = pd.read_parquet("../data/processed/karachi_events_clean.parquet")
lab = df[df.actor_known].copy()
print("labelled events:", len(lab), "| groups:", lab.gname.nunique())

vc = lab.gname.value_counts()
print(vc.head(12).to_string())
for m in [5, 10, 15, 20]:
    keep = vc[vc >= m]
    print(f"min {m:2}: {len(keep):2} groups cover {keep.sum():3} events ({keep.sum()/len(lab):.0%})")

# twin records: how many labelled events share a dup_group
print("labelled events in a shared dup_group:", lab.dup_group.duplicated(keep=False).sum())

# label noise: attributions GTD itself marks as suspected
print("uncertain attributions:", int((lab.guncertain1 == 1).sum()))

# leakage: does the summary name the group, or talk about attribution?
name_in_text = [g.lower() in s.lower() for g, s in zip(lab.gname, lab.summary_clean)]
print("summary contains the full group name:", f"{sum(name_in_text)/len(lab):.0%}")
words = "attributed|claimed responsibility|blamed|suspected|believed to"
print("summary mentions attribution wording:",
      f"{lab.summary_clean.str.contains(words, case=False).mean():.0%}")
```
