```python
import pandas as pd
pd.set_option("display.max_colwidth", 220)
df = pd.read_parquet("../data/processed/karachi_events_clean.parquet")

# 1. What the location text looks like
print("Share with location text:", df["location"].notna().mean().round(2))
print(df["location"].dropna().sample(15, random_state=2).to_string())
```

```python
# 2. The opening phrases (templates) of the location text
starts = df["location"].dropna().apply(lambda s: " ".join(s.split()[:5]))
print(starts.value_counts().head(15))
```

```python
# 3. Events with no location text: what does the summary say instead?
print(df.loc[df["location"].isna(), "summary_clean"].sample(8, random_state=2).to_string())
```

```python
# 4. District names already recorded in the original city field
print(df.loc[df["city_raw"].str.lower() != "karachi", "city_raw"].value_counts().to_string())
```

```python
# 5. How often do the texts use common locality words?
words = r"\b(area|neighbou?rhood|town|locality|colony|sector|chowrangi|road|market)\b"
for col in ["location", "summary_clean"]:
    print(col, df[col].str.contains(words, case=False, na=False).mean().round(2))
```
