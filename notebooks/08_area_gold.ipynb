```python
import pandas as pd
df = pd.read_parquet("../data/processed/karachi_events_clean.parquet")

with_loc = df[df["location"].notna()].sample(40, random_state=42)
without_loc = df[df["location"].isna()].sample(20, random_state=42)

gold = pd.concat([with_loc, without_loc])[["eventid", "location", "summary_clean"]].copy()
gold["gold_area"] = ""
gold.to_csv("../data/processed/area_gold_blank.csv", index=False)
print(len(gold), "rows saved")
```

```python
import pandas as pd
df = pd.read_parquet("../data/processed/karachi_events_clean.parquet")
used = pd.read_csv("../data/processed/area_gold.csv")
used_keys = set(used["summary_clean"].map(lambda s: " ".join(str(s).split())))
fresh = df[~df["summary_clean"].map(lambda s: " ".join(str(s).split())).isin(used_keys)]

test = pd.concat([
    fresh[fresh["location"].notna()].sample(20, random_state=7),
    fresh[fresh["location"].isna()].sample(10, random_state=7),
])[["location", "summary_clean"]].copy()
test["gold_area"] = ""
test.to_csv("../data/processed/area_test_blank.csv", index=False)
```
