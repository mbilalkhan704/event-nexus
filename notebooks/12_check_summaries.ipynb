```python
import pandas as pd
from sentence_transformers import SentenceTransformer

df = pd.read_parquet("../data/processed/karachi_events_clean.parquet")
model = SentenceTransformer("../models/minilm-embedder")

texts = df["summary_clean"].fillna("")
lens = texts.map(lambda t: len(model.tokenizer.tokenize(t)))

print(model.max_seq_length)
print(lens.describe())
print("empty:", (texts.str.strip() == "").sum())
print("over limit:", (lens > model.max_seq_length).sum())
```
